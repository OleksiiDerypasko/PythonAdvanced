**Кожну вправу розв’язати двома способами**: (NumPy та ітеративно) і порівняти час

```python
import numpy as np

import random

import time
from functools import wraps

def time_it(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        end = time.perf_counter()
        elapsed = end - start
        return result, elapsed
    return wrapper

def print_matrix(data):
    if isinstance(data, np.ndarray):
        data = data.tolist()

    for row in data:
        for elem in row:
            print(f"{elem:2}", end=" ")
        print()
    print()
```

**Вправа 3.** Ввести 4 числа \(n, m, r, c\). Побудувати матрицю розміру \(n \times m\), у якій кожний \(r\)-й рядок і кожний \(c\)-й стовпчик дорівнюють 0, інші елементи — 1.

```python
@time_it
def task_3_np(n, m, r, c):
    arr_np = np.ones((n, m), dtype=int)
    arr_np[::r, :] = 0
    arr_np[:, ::c] = 0
    return arr_np

@time_it
def task_3_iter(n, m, r, c):
    arr_iter = []
    for i in range(n):
        row = []
        for j in range(m):
            if i % r == 0 or j % c == 0:
                row.append(0)
            else:
                row.append(1)
        arr_iter.append(row)
    return arr_iter


res_np, time_np = task_3_np(5, 10, 2, 3)
res_iter, time_iter = task_3_iter(5, 10, 2, 3)

print(f"Numpy (час - {time_np})")
print_matrix(res_np)

print(f"Iter (час - {time_iter})")
print_matrix(res_iter)

res_np, time_np = task_3_np(n, m, 2, 3)
res_iter, time_iter = task_3_iter(n, m, 2, 3)
print(f"На великих значеннях:\nNumpy (час - {time_np})\nIter (час - {time_iter})")
print(f"Співпадіння результатів: {np.allclose(res_np, res_iter)}\n")
```

**Вправа 10.** Ввести \(n, m\). Згенерувати матрицю випадкових значень \(n \times m\) на \((0,1)\); знайти мінімум, максимум, середнє та стандартне відхилення (округлити до 3 знаків).

```python
@time_it
def task_10_np(arr_np):
    min_val = arr_np.min()
    max_val = arr_np.max()
    mean_val = arr_np.mean()
    std_val = arr_np.std()
    return (min_val, max_val, mean_val, std_val)


@time_it
def task_10_iter(data):
    n = len(data)
    m = len(data[0]) if n > 0 else 0
    count = n * m

    i = 0
    total_sum = 0.0
    min_val = float("inf")
    max_val = float("-inf")

    while i < n:
        j = 0
        while j < m:
            v = data[i][j]
            total_sum = total_sum + v
            if v < min_val:
                min_val = v
            if v > max_val:
                max_val = v
            j = j + 1
        i = i + 1

    mean_val = total_sum / count

    i = 0
    sum_sq = 0.0
    while i < n:
        j = 0
        while j < m:
            d = data[i][j] - mean_val
            sum_sq = sum_sq + d * d
            j = j + 1
        i = i + 1
    std_val = (sum_sq / count) ** 0.5

    return (min_val, max_val, mean_val, std_val)

n, m = 5, 5
arr_np = np.random.rand(n, m)
data = arr_np.tolist()
print_matrix(data)

res_np, time_np = task_10_np(arr_np)
res_iter, time_iter = task_10_iter(data)

print(f"Numpy (час - {time_np})")
print(res_np)

print(f"Iter (час - {time_iter})")
print(res_iter)

n, m = 1000, 1000
arr_np = np.random.rand(n, m)
data = arr_np.tolist()

res_np, time_np = task_10_np(arr_np)
res_iter, time_iter = task_10_iter(data)

print(f"На великих значеннях:\nNumpy (час - {time_np})\nIter (час - {time_iter})")
print(f"Співпадіння результатів: {np.allclose(res_np, res_iter)}\n")
```

## Завдання 2. Розв’язання системи лінійних алгебраїчних рівнянь (СЛАР) за формулами Крамера

### Постанова задачі
Розв’язати систему з чотирма невідомими $x_1,x_2,x_3,x_4$ за **формулами Крамера**, а також виконати перевірку результатів трьома способами:
1) Матричним множенням ($A\,x \approx b$);
2) Через обернену матрицю ($x = A^{-1}b$);
3) За допомогою `numpy.linalg.solve()`.

**Порівняти** всі розв’язки функцією `numpy.allclose()`.

---

### Вхідні дані

Система має вигляд:
$$
\begin{cases}
x_1 + 2x_2 + 7x_3 + 5x_4 = 16,\\
0\cdot x_1 + 7x_2 + 5x_3 + 3x_4 = 4,\\
7x_1 + 5x_2 + 3x_3 + 0\cdot x_4 = 0,\\
x_1 + 2x_2 + 7x_3 + 5x_4 = 12.
\end{cases}
$$

Матриця коефіцієнтів і вектор правих частин:
$$
A=
\begin{bmatrix}
1&2&7&5\\
0&7&5&3\\
7&5&3&0\\
1&2&7&5
\end{bmatrix},
\qquad
b=
\begin{bmatrix}
16\\
4\\
0\\
12
\end{bmatrix}.
$$

---

### Короткий опис алгоритму (Крамер)

Нехай $A x = b$ і $D=\det(A)\neq 0$.
1. Обчислити $D=\det(A)$.
2. Для кожного $j=1,\dots,4$ сформувати $A_j$ — матрицю $A$ з заміною $j$-го стовпця на $b$.
3. Обчислити $D_j=\det(A_j)$.
4. Знайти компоненти розв’язку: $x_j=\dfrac{D_j}{D}$.

---
### Перевірки коректності:
- $A\,x \approx b$ (матричне множення),
- $x_{\text{inv}} = A^{-1}b$ (обернена матриця),
- $x_{\text{solve}}=\texttt{numpy.linalg.solve}(A,b)$,
- порівняння: `np.allclose(x, x_inv)` і `np.allclose(x, x_solve)`.


```python
import numpy as np

A = np.array([
    [1, 3, 5, 7],
    [3, 5, 7, 1],
    [5, 7, 1, 3],
    [7, 1, 3, 5]
], dtype=float)

b = np.array([12, 0, 4, 16], dtype=float)


def cramers_rule(A_in, b_in):
    A = np.asarray(A_in, dtype=float)
    b = np.asarray(b_in, dtype=float)

    if A.ndim != 2:
        raise ValueError("A має бути матрицею 2D.")
    n, m = A.shape
    if n != m:
        raise ValueError("A має бути квадратною.")
    if b.ndim != 1:
        raise ValueError("b має бути вектором 1D.")
    if b.shape[0] != n:
        raise ValueError("Розмірність b не відповідає A.")

    detA = np.linalg.det(A)
    if np.isclose(detA, 0.0):
        raise ValueError("det(A) = 0 → немає єдиного розв'язку (Крамера не застосувати).")

    x = np.zeros(n, dtype=float)

    j = 0
    while j < n:
        Aj = A.copy()
        Aj[:, j] = b
        detAj = np.linalg.det(Aj)
        x[j] = detAj / detA
        j = j + 1

    return x, detA


x_cramer, detA = cramers_rule(A, b)

A_inv = np.linalg.inv(A)
x_inv = A_inv @ b

x_solve = np.linalg.solve(A, b)

check_cramer = np.allclose(A @ x_cramer, b)
check_inv    = np.allclose(A @ x_inv,    b)
check_solve  = np.allclose(A @ x_solve,  b)

cmp_ci = np.allclose(x_cramer, x_inv)
cmp_cs = np.allclose(x_cramer, x_solve)
cmp_is = np.allclose(x_inv,    x_solve)

print("det(A) =", detA)

print("x (Крамер):")
i = 0
while i < x_cramer.shape[0]:
    print("  x[", i, "] =", x_cramer[i])
    i = i + 1

print("x (Inverse):")
i = 0
while i < x_inv.shape[0]:
    print("  x[", i, "] =", x_inv[i])
    i = i + 1

print("x (np.linalg.solve):")
i = 0
while i < x_solve.shape[0]:
    print("  x[", i, "] =", x_solve[i])
    i = i + 1

print("Перевірки A @ x ≈ b:")
print("  Крамер :", check_cramer)
print("  Inverse:", check_inv)
print("  Solve  :", bool(check_solve))

print("Порівняння розв'язків:")
print("  Співпаднння результатів (Крамер, Inverse) =", cmp_ci)
print("  Співпаднння результатів (Крамер, Solve)   =", cmp_cs)
print("  Співпаднння результатів (Inverse, Solve)  =", cmp_is)
```

- Час NumPy-варіанту, як правило, **менший**, ніж у ітеративного.


```python
A = np.array([[5, 2, 0],
              [10, 4, 1],
              [7, 3, 2]], dtype=int)

B = np.array([[ 3,  6, -1],
              [-1, -2,  0],
              [ 2,  1,  3]], dtype=int)

# (A - B @ B) @ (2*A + B)
@time_it
def expr_numpy(A, B):
    BB = B @ B
    left = A - BB
    right = (2 * A) + B
    res = left @ right
    return res

def mat_shape(M):
    n = len(M)
    if n == 0:
        m = 0
    else:
        m = len(M[0])
    return n, m

def zeros_like(n, m):
    Z = []
    i = 0
    while i < n:
        row = []
        j = 0
        while j < m:
            row.append(0)
            j = j + 1
        Z.append(row)
        i = i + 1
    return Z

def mat_add(X, Y):
    n, m = mat_shape(X)
    Z = zeros_like(n, m)
    i = 0
    while i < n:
        j = 0
        while j < m:
            Z[i][j] = X[i][j] + Y[i][j]
            j = j + 1
        i = i + 1
    return Z

def mat_sub(X, Y):
    n, m = mat_shape(X)
    Z = zeros_like(n, m)
    i = 0
    while i < n:
        j = 0
        while j < m:
            Z[i][j] = X[i][j] - Y[i][j]
            j = j + 1
        i = i + 1
    return Z

def scalar_mul(k, X):
    n, m = mat_shape(X)
    Z = zeros_like(n, m)
    i = 0
    while i < n:
        j = 0
        while j < m:
            Z[i][j] = k * X[i][j]
            j = j + 1
        i = i + 1
    return Z

def mat_mul(X, Y):
    n, m = mat_shape(X)
    p, q = mat_shape(Y)
    if m != p:
        raise ValueError("Розмірності несумісні для множення.")
    Z = zeros_like(n, q)
    i = 0
    while i < n:
        j = 0
        while j < q:
            s = 0
            k = 0
            while k < m:
                s = s + X[i][k] * Y[k][j]
                k = k + 1
            Z[i][j] = s
            j = j + 1
        i = i + 1
    return Z

A_list = A.tolist()
B_list = B.tolist()

@time_it
def expr_iter(A_list, B_list):
    B2 = mat_mul(B_list, B_list)
    left = mat_sub(A_list, B2)
    twoA = scalar_mul(2, A_list)
    right = mat_add(twoA, B_list)
    res = mat_mul(left, right)
    return res

res_np, t_np = expr_numpy(A, B)
res_it, t_it = expr_iter(A_list, B_list)

print("NumPy результат:")
print_matrix(res_np)
print("Ітеративний результат:")
print_matrix(np.array(res_it))

print(f"Співпадіння результатів: {np.allclose(res_np, res_it)}\n")

print(f"NumPy:      {t_np:.6f} c")
print(f"Ітеративно: {t_it:.6f} c")
```
