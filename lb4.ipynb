## Хід виконання

Реалізуємо функцію яка стане основою декоратора. Декоратор буде приймати аргумент у якості цілого числа ліміту повторень (додана перевірка на натуральність числа). У тілі декоратору реалізована функціональна частина `deco(func)` з аргументом функції, яка буде декорованою. У функціональній частині-обгортки декоратору реалізовані додаткові методи-обгортки, які приймають `*args` та `**kwargs` та вже в їх тілі оброблена обробка перевірки змінної (кіл-ті ітерацій). Щоб змінна лічильника вже пройдених операцій `calls` працювала коректно, вона оголошується у тілі декоратора `deco`, а в обгорці реалізоваються глобальна ініціалізація `nonlocal` та маніпуляція `calls += 1`

```python
from functools import wraps

def limit_calls(limit=3):
    def deco(func):
        calls = 0
        @wraps(func)
        def wrapper(*args, **kwargs):
            nonlocal calls
            if calls >= limit:
                print(f"{func.__name__} can be called only {limit} times")
                return
            calls += 1
            return func(*args, **kwargs)
        return wrapper
    return deco
```

Реалізовуємо тестові методи для перевірки з різною кіл-тю аргументів і типів:
- `selfname` немає агрументів і виводить власну назву (для цього був доданий в тілі декоратору вбудований декоратор `@wraps`)
- `sum` виводить суму двох аргументів (кіл-ть аргументів константна та вказаний власний ліміт `2`)
- `power` повертає степінь числа (вимагається один аргумент, проте може передаватися конкретне значення)
- `print_sum_array` повертає суму лише цілочисельних елементів масиву (аргумент - один масив)

```python
import sys

@limit_calls()
def selfname():
    print(selfname.__name__)

@limit_calls(2)
def sum(a, b):
    print(a + b)

@limit_calls()
def power(x, exp=2):
    print(x ** exp)

@limit_calls()
def print_sum_array(arr):
    sum = 0
    for i in arr:
        if type(i) == int:
            sum += i
    print(sum)
```

## Тестування

```python
print("===Тестування методу selfname (limit = 3)===")
for i in range(4):
    selfname()

print("\n===Тестування методу sum (limit = 2)===")
numbers = [1,2,3,4,5,6,7,8]
for i in range(len(numbers)//2):
    sum(numbers[i], numbers[i+3])

print("\n===Тестування методу power (limit = 3)===")
for i in range(len(numbers)//2):
    if i%2 != 0:
        power(numbers[i], numbers[i+3])
    else:
        power(numbers[i])

print("\n===Тестування методу print_sum_array (limit = 3)===")
arrays = [["str", 1, 2, -9],
          ["with float", 5, [], 1.5, sys.path],
          [2, 4],
          []]

for array in arrays:
    print_sum_array(array)
```
